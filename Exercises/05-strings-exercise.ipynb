## 1. Counting letters (*)

Let the user input a word: 

&nbsp; a) &nbsp; print out the number of letters in the word. (*) 
  
&nbsp; b) &nbsp; print out the number of **uppercase** and **lowercase** letters of the word (*)   


````python
# A
user_input = input("Input a word of your choice: ")

print(f"The number of letters in the word you wrote is: {len(user_input)}")
````

````python
# B
user_input = input("Input a word of your choice: ")

upper_case = sum(1 for letter in user_input if letter.isupper())
lower_case = sum(1 for letter in user_input if letter.islower())

print(f"The number of upper case letters in the word you wrote is {upper_case} and the number of lower case letters is {lower_case}!")
````

## 2. Counting words (*)

Count the number of words in this sentence: "A picture says more than a thousand words, a matematical formula says more than a thousand pictures." (*)

<summary>Answer</summary>

## 3. Palindrome (*)

A palindrome is a sequence of characters that is the same, when read forward as backwards (ignoring spaces). For example:
- "Anna" is a palindrome
- "Ni talar bra latin"
- bjkjb 

Let the user input a sequence of characters and check if it is a palindrome. (*)

````python
user_input = input("Input a palindrome (or something you think is a palindrome): ")

if user_input == user_input[::-1]:
    print(f"The word {user_input} is a palindrome!")
else:
    print(f"The word {user_input} is not a palindrome!")
````

## 4. Vowels (**)

Count the number of vowels in this sentence: "Pure mathematics is, in its way, the poetry of logical ideas"

<details>

<summary>Answer</summary>

```

There are 22 vowels in this sentence

```
</details>

````python
sentence = "Pure mathematics is, in its way, the poetry of logical ideas"
vokaler = "aeiouyåäö"

vowels = [1 for letter in sentence.lower() if letter in vokaler]

print(f'"{sentence}"')
print(f"The number of vowels in the sentence above is {len(vowels)}.")
````

## 5. Encryption (**)

Let the user input a word and: 

&nbsp; a) &nbsp; encrypt the message by replacing each letter with the next letter. If the letter is in the end of the alphabet, use the first letter instead. 

e.g. in Swedish: "höst" $\rightarrow$ "iatu"

&nbsp; b) &nbsp; decrypt the message

&nbsp; c) &nbsp; let the user choose either encryption or decryption.

````python
user_input = input("Please input a word: ")
split_input = list(user_input.lower())
swedish_alphabet = "abcdefghijklmnopqrstuvwxyzåäö"

encrypted_input = []

# A
for letter in split_input:
    if letter in swedish_alphabet:
        id = swedish_alphabet.index(letter)
        next_letter = swedish_alphabet[(id+1) % len(swedish_alphabet)]
        encrypted_input.append(next_letter)
    else:
        encrypted_input.append(letter)

encrypted_output = "".join(encrypted_input)
    
print(f'The encrypted word for "{user_input}" is "{encrypted_output}".')
````

````python
# B, alt. 1
swedish_alphabet = "abcdefghijklmnopqrstuvwxyzåäö"

decrypted_input = []

for letter in encrypted_output:
    if letter in swedish_alphabet:
        id = swedish_alphabet.index(letter)
        previous_letter = swedish_alphabet[(id-1) % len(swedish_alphabet)]
        decrypted_input.append(previous_letter)
    else:
        decrypted_input.append(letter)

decrypted_output = "".join(decrypted_input)

print(f'The decrypted word for "{encrypted_output}" is "{decrypted_output}".')
````

````python
# B, alt. 2
swedish_alphabet = "abcdefghijklmnopqrstuvwxyzåäö"

def encoder_decoder(message, choice):
    output = []

    for letter in message:
        if letter in swedish_alphabet:
            id = swedish_alphabet.index(letter)
            new_letter = swedish_alphabet[(id + choice) % len(swedish_alphabet)]
            output.append(new_letter)
        else:
            output.append(letter)

    return "".join(output)

encrypted_input = encoder_decoder(split_input, 1)
decrypted_output = encoder_decoder(encrypted_input, -1)

print(f'The decrypted word is "{decrypted_output}" and the encrypted version of the word is "{encrypted_input}".')
````
