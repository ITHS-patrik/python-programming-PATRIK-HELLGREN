## 2. Food menu (*)

Create a:  

&nbsp; a) &nbsp; **list** with the following elements: "vegetarisk lasagne", "spaghetti", "fisk", "grönsakssoppa", "pannkakor". (*)

&nbsp; b) &nbsp; **list** with the weekdays (*)

&nbsp; c) &nbsp; create a food menu with each day corresponding to each food item and print it out. (*)

<details>

<summary>Answer</summary>

c) 
```
Bambameny
Mån: vegetarisk lasagne
Tis: spaghetti
Ons: fisk

```

</details>

````python
# A
menu = ["Vegetarisk lasagne", "Spaghetti", "Fisk", "Grönsakssoppa", "Pannkakor"]
````

````python
# B
weekdays = ["Måndag", "Tisdag", "Onsdag", "Torsdag", "Fredag"]
````

````python
# C FOR LOOP (DICT)
print("Bambameny:")
for i in range(len(weekdays)):
    week_menu = {weekdays[i]: menu[i]}
    print(week_menu)

# C LIST COMPREHENSION (clean)
max_letters_in_weekday = max(len(day) for day in weekdays)

for day, food in zip(weekdays, menu):
    print(day.ljust(max_letters_in_weekday), ":", food)
````

````python
# A
import random

dice_rolls = []

for i in range(100):
    dice_rolls.append(random.randint(1,6))
    
sum_of_six = dice_rolls.count(6)
print(f'The number of 6 within 100 dice throws was {sum_of_six}.')
````

````python
# B
#random.seed(1) # Samma "slumpmässiga" nummer varje gång. Inaktivera för att slumpa fram nya nummer.
dice_throws = [10, 100, 1000, 10000, 100000, 1000000]
amount_of_six = []
probability = []

for throws in dice_throws:
    count = 0
    for i in range(throws):
        if random.randint(1, 6) == 6:
            count += 1
    amount_of_six.append(count)
    probability.append(count / throws)

print(amount_of_six)
print(probability)
````

## 6. Monte Carlo simulation (**)

&nbsp; a) Simulate 5000 uniformly distributed points between -1 and 1, $X\sim U(-1,1)$. Now calculate the Euclidean distances between each point and the origin (0,0), $d_i = \sqrt{x_i^2 + y_i^2}$. For $d_i < 1$ plot the point with one color, and for $d_i > 1$ plot the point in another color, $i = \{1,\ldots, 5000\}$.  (**)

&nbsp; b) Calculate the fraction between number of inner points and number of outer points. What do you this will converge to when number of points reaches infinity? (**)

````python
# A

import random
import math
import matplotlib.pyplot as plt

x = []
y = []
data_points = 5000 # Change to e.g. 150 000 for exercise B to see the effects.

for i in range(data_points):
    x.append(random.uniform(-1,1))
    y.append(random.uniform(-1,1))

distance = [math.sqrt(x[i]**2 + y[i]**2) for i in range(data_points)]

x_lower = [x[i] for i, d in enumerate(distance) if d <= 1]
y_lower = [y[i] for i, d in enumerate(distance) if d <= 1]

x_higher = [x[i] for i, d in enumerate(distance) if d > 1]
y_higher = [y[i] for i, d in enumerate(distance) if d > 1]

plt.scatter(x_lower, y_lower, marker="o", color="red", label="distance < 1 from (0, 0)", alpha=0.4)
plt.scatter(x_higher, y_higher, marker="o", color="blue", label="distance > 1 from (0, 0)", alpha=0.4)

plt.title("Monte Carlo simulation")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc="center")
plt.show()
````

````python
# B
fraction = len(x_lower) / data_points
print(f"The fraction of data points with a distance of less than 1 from (0,0) is {fraction}.")

print(f"I believe that an infinite amount of data points will converge towards the actual probability of the randomized data point falling within the circle ({fraction:.1%} in this case).")
````

````python
# A
import random

first_door = random.randint(1,3)
print(f"First door picked is {first_door}.")

not_third_door = [1,2,3]
not_third_door.remove(first_door)

if first_door == 1:
    second_door = random.randint(2,3)
    not_third_door.remove(second_door)
elif first_door == 2:
    second_door = random.choice([1,3])
    not_third_door.remove(second_door)
else:
    second_door = random.randint(1,2)
    not_third_door.remove(second_door)

print(f"The door that opens is {second_door}.")

third_door = not_third_door[0]
print(f"(The third door is door number {third_door}.)")

chosen_door = random.choice([first_door,third_door])
print(f"The door chosen is door number {chosen_door}.")

print()
print("WOULD I CHOOSE TO STAY WITH THE INITIALLY CHOSEN DOOR OR CHANGE TO THE OTHER ONE? 50/50 CHANCE..?")
# After doing Exercise B, I can confirm I was wrong above (which I kinda knew since it would've been too easy.)
# See exercise B below for the answer.

````

````python
# B, prints
import random

doors = [1,2,3]
doors_left = [1,2,3]

# FIRST DOOR = EITHER SNAKE OR RABBIT
first_door = random.randint(1,3)
doors_left.remove(first_door) # underlag för second_door där en av ormarna finns
print(f"First door picked is door number {first_door}.")

# SECOND_DOOR = SNAKE
if first_door == 1:
    second_door = random.choice(doors_left)
    doors.remove(second_door) # remove snake door 1 out of 2.
elif first_door == 2:
    second_door = random.choice(doors_left)
    doors.remove(second_door) # remove snake door 1 out of 2.
else:
    second_door = random.choice(doors_left)
    doors.remove(second_door) # remove snake door 1 out of 2.

print(f"The first door that opens is door number {second_door}. WAH - a snake!")

# RABBIT
rabbit = random.choice(doors)
print(f"The rabbit is in door number {rabbit}.")
print(f"Choosing between door number {doors[0]} and {doors[1]}.")
doors.remove(rabbit)

# THIRD DOOR = EITHER SNAKE OR RABBIT
third_door = doors[0]

chosen_door = random.choice([first_door,third_door])
print(f"The door chosen is door number {chosen_door}.")

if chosen_door == rabbit:
    print("The rabbit was found!")
else:
    print("The rabbit was NOT found!")
````

````python
# B, plot
import random
import matplotlib.pyplot as plt

simulations = [10, 100, 1000, 10000, 100000, 1000000]
stay_probability = []
switch_probability = []

for sim in simulations:
    stay_count = 0
    switch_count = 0

    for _ in range(sim):

        # Rabbit door
        rabbit = random.randint(1,3)

        # First door to be picked (but not opened)
        first_door = random.randint(1,3)
        doors = [1,2,3]
        doors.remove(first_door)
        if rabbit in doors: # Remove rabbit if it's still in doors. Leaves one or two doors.
            doors.remove(rabbit)
        
        # Second door (that opens) -> only snake door(s) left
        second_door = random.choice(doors)

        # Switch door
        doors = [1,2,3]
        doors.remove(first_door)
        doors.remove(second_door)
        switch_door = doors[0]

        if first_door == rabbit:
            stay_count += 1
        if switch_door == rabbit:
            switch_count += 1
        
    stay_probability.append(stay_count / sim)
    switch_probability.append(switch_count / sim)

print(f"Stay probability: {stay_probability}")
print(f"Switch probability: {switch_probability}")
print(f"You have a {switch_probability[-1]} chance of winning if you switch door!")

plt.plot(simulations,stay_probability, marker="o", color="red", label="Stay probability")
plt.plot(simulations,switch_probability, marker="o", color="green", label="Switch probability")
plt.title("Doors: rabbit & snakes")
plt.xlabel("Simulations")
plt.ylabel("Probability")
plt.grid(True)
plt.xscale("log")
plt.legend()
plt.show()
````
