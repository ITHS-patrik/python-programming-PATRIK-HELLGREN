## 1. Dice rolls (*)

## 2. Food menu (*)

Create a:  

&nbsp; a) &nbsp; **list** with the following elements: "vegetarisk lasagne", "spaghetti", "fisk", "grönsakssoppa", "pannkakor". (*)

&nbsp; b) &nbsp; **list** with the weekdays (*)

&nbsp; c) &nbsp; create a food menu with each day corresponding to each food item and print it out. (*)

<details>

<summary>Answer</summary>

c) 
```
Bambameny
Mån: vegetarisk lasagne
Tis: spaghetti
Ons: fisk
Tor: grönsakssoppa
Fre: pannkakor
```

</details>

````python
import pprint

# A
menu = ["Vegetarisk lasagne", "Spaghetti", "Fisk", "Grönsakssoppa", "Pannkakor"]
````

````python
# B
weekdays = ["Måndag", "Tisdag", "Onsdag", "Torsdag", "Fredag"]
````

````python
# C FOR LOOP (DICT)
print("Bambameny:")
for i in range(len(weekdays)):
    weekMenu = {weekdays[i]: menu[i]}
    print(weekMenu)

# C LIST COMPREHENSION
maxLenWeekdays = max(len(day) for day in weekdays)

for day, food in zip(weekdays, menu):
    print(day.ljust(maxLenWeekdays), ":", food)
````

## 5. Dice rolls convergence (**)

Simulate: 

&nbsp; a) &nbsp; 100 dice rolls and count the number of outcome six. (*)

&nbsp; b) &nbsp; 10, 100, 1000, 10000, 100000, 1000000 dice rolls. Count the number of outcome six in each simulation and store it in a list. Compute the probability of outcome six in each simulation. (**)

&nbsp; c) &nbsp; Use matplotlib to plot this list. (*)

<details>

<summary>Answer </summary>

```

b) 
```
Count number of six: [1, 21, 161, 1642, 16560, 166406]

c) Note that the graph changes due to random simulation. To reproduce this result, use

<img align="left" src="../assets/probability_six_graph.png" width="300"/>

</details>

````python
# A
import random

diceRolls = []

for i in range(100):
    diceRolls.append(random.randint(1,6))
    
sumOfSix = diceRolls.count(6)
print(f'The number of 6 within 100 dice throws was {sumOfSix}.')
````

````python
# B
#random.seed(1) # Samma "slumpmässiga" nummer varje gång. Inaktivera för att slumpa fram nya nummer.
diceThrows = [10, 100, 1000, 10000, 100000, 1000000]
amountOfSix = []
probability = []

for throws in diceThrows:
    count = 0
    for i in range(throws):
        if random.randint(1, 6) == 6:
            count += 1
    amountOfSix.append(count)
    probability.append(count / throws)

print(amountOfSix)
print(probability)
````

````python
# C
random.seed(1)
plt.plot(diceThrows, probability, marker="o", color="red", label="Actual probability")
plt.title("Probability of six")
plt.xlabel("Dice throws")
plt.ylabel("Probability")
plt.axhline(y=1/6, color="green", linestyle="--", label="Theoretical probability")
plt.legend() # måste vara efter axhline, annars syns inte <label>.
plt.grid(True)
plt.xscale("log")
plt.show()
````
