## 2. Euclidean distance (*)

The formula for Euclidean distance in 2D between $P: (p_1, p_2)$ and $Q: (q_1, q_2)$  is: 

$d(P,Q) = \sqrt{(p_1-q_1)^2 + (p_2-q_2)^2}$

&nbsp; a) &nbsp; Create a function that takes two points as input parameters and return the Euclidean between them. (*)

&nbsp; b) &nbsp; Let the user input two points. Call the function using the users input points. (*)

&nbsp; c) &nbsp; Use your function to calculate distances between the origin (0, 0) and each of these points: (10, 3), (-1, -9), (10, -10), (4, -2), (9, -10). (*)

<details>

<summary>Hint</summary>

Use a **for** statement

</details>
<br>
<details>

<summary>Answer</summary>

c)

```
[10.4, 9.1, 14.1, 4.5, 13.5]
```

</details>


````python
import math

# A
point1 = (21, 35)
point2 = (15, 23)

def euclidean_distance(p1, p2):

    d0 = p1[0]-p2[0]
    d1 = p1[1]-p2[1]
    return round(math.sqrt(d0**2+d1**2), 1)

distance = euclidean_distance(point1, point2)
print(distance)
````

````python
# B
x1, y1 = input("Please input the first point separated by a comma (x, y): ").split(",")
input1 = (int(x1.strip()), int(y1.strip()))

x2, y2 = input("Please input the second point separated by a comma (x, y): ").split(",")
input2 = (int(x2.strip()), int(y2.strip()))

euclidean_distance(input1, input2)
````

````python
# C
coordinates = [(10, 3), (-1, -9), (10, -10), (4, -2), (9, -10)]
origin = (0, 0)
distances = []

for i, coordinate in enumerate(coordinates):

    distances.append(euclidean_distance(coordinate, origin))

print(f"{distances}")
````

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10)

def calculate_line(x, m):
    
    line = x + m
    
    return line

f_line = calculate_line(x * x, -3)
g_line = calculate_line(4 * x, -7)

plt.plot(x, f_line, color="blue")
plt.plot(x, g_line, color="orange")
plt.show()

# Linjerna möts i en punkt. Punkten räknas ut genom:
# x ** 2 - 3 = 4 * x - 7
# x = 2
# T ex 2 ** 2 - 3 = 1 (y)
# Linjerna skär varandra i punkt: (2, 1)
# Linjen g(x) är en tangent till parabeln f(x) när x = 2.
````

- removes all leading and trailing blank spaces

````python
names = ["  MaRcUs ", " iDA aNderSon", "OLOF Olofsson            "]

def clean_name(name):
    
    stripped = name.strip().title()
    print(f"{stripped}")    

for name in names:
    clean_name(name)
````

## 5. Change (**)

Create a function that takes a value as input parameter and print out the banknotes and coins in Swedish currency representing this value. For example 5289 would give the following printout: 

- 5st 1000-lapp
- 1st 200-lapp
- 1st 50-lapp
- 1st 20-lapp
- 1st 10-krona
- 1st 5-krona
- 2st 2-krona

Now let the user input a value, and use the function to calculate the change.

````python
import math
banknotes = [1000, 500, 200, 100, 50, 20, 10, 5, 2, 1]

def change2(value):
    for banknote in banknotes:
        quotient, rest = divmod(value, banknote)
        value = rest

        if quotient != 0:
            print(f"{quotient}st {banknote} SEK")

value = int(input("Please input a sum of SEK (numbers only): "))
change2(value)
````
