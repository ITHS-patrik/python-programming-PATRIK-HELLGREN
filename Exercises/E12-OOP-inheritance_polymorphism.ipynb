## 1. Video (*)

Create classes following this UML:

<img src="../assets/UML_video_polymorphism.png" width="700"/>

Note that the method info() should be different in the different classes where it should be implemented. 

Use the following code to test your program.

```python
pokemon = TV_serie("Pokemon", "Cartoon", 4.5, 550)
titanic = Movie("Titanic", "Romance", 4.7, 194)
code = Documentary("The Code", "Math", 4)

for video in tuple((pokemon, titanic, code)):
    print(video.info())
```

An example output could be: 
```
TV series with title Pokemon, genre Cartoon, rating 4.5 and episodes 550

Movie with title Titanic, genre Romance, rating 4.7, duration 194 minutes

Video with title The Code, genre Math and rating 4
```

(*)

<details>

<summary>Hint</summary>

Use ```__super__()``` in each subclass to call the \_\_init\_\_() in parent class. Add additional parameters in the \_\_init\_\_() of each subclass when needed. Keep error handling and validation in parent class and let the subclass inherit them in order to avoid repeating validation code.

</details>

````python
class Video:

    def __init__(self, title: str, genre: str, rating: float):
        self.title = title
        self.genre = genre
        self.rating = rating

    def info(self):
        return f"Video with title {self.title}, genre {self.genre} and rating {self.rating}"
    
class TV_serie(Video):

    def __init__(self, title, genre, rating, num_episodes: int):
        super().__init__(title, genre, rating)

        self.num_episodes = num_episodes

    def info(self):
        return f"TV series with title {self.title}, genre {self.genre}, rating {self.rating} and episodes {self.num_episodes}"
    
class Movie(Video):

    def __init__(self, title, genre, rating, duration: float):
        super().__init__(title, genre, rating)

        self.duration = duration

    def info(self):
        return f"Movie with title {self.title}, genre {self.genre}, rating {self.rating} and duration {self.duration} minutes"
    
class Documentary(Video):

    def __init__(self, title, genre, rating):
        super().__init__(title, genre, rating)

pokemon = TV_serie("Pokemon", "Cartoon", 4.5, 550)
titanic = Movie("Titanic", "Romance", 4.7, 194)
code = Documentary("The Code", "Math", 4)

for video in tuple((pokemon, titanic, code)):
    print(video.info())
````

## 2. Fraction (**)

Create a class called Frac to represent mathematical fractions. The class is instantiated with two instance variables: nominator and denominator. Objects instantiated from this class should have methods for addition, subtraction, multiplication, division using the operators +,-,*,/. Note that these implemented methods must be mathematically correct. Also implement the following methods: 

```python

simplify(self, value = None) # simplifies to most simple form unless value is given 

__str__(self) # represent the fraction in a neat way for printing

mixed(self) # represent the fraction in mixed terms 

__eq__(self, other) # checks equality by overloading ==

```

Also remember to handle errors and validations.

Example of tests that it should handled: 

- 1/2 + 1/3 = 5/6
- 1/2 - 1/3 = 1/6
- 7/6 --> 1 1/6 (mixed)
- 3*1/2 = 3/2
- 1/2 * 3 = 3/2
- 1/4 + 2 = 9/4
- 1/4 / 1/2 = 1/2
- 2/4 == 1/2 --> True
- 3/4 += 2 = 11/4

</details>


````python
from math import gcd

class Frac:

    def __init__(self, nominator, denominator):
        self.nominator = nominator
        self.denominator = denominator

    def __add__(self, other):
        if isinstance(other, int):
            other = Frac(other, 1)
        if isinstance(other, Frac):
            new_nom = self.nominator * other.denominator + other.nominator * self.denominator
            new_den = self.denominator * other.denominator
            result = Frac(new_nom, new_den)
            return result.mixed()
        return NotImplemented
        
    def __radd__(self, other):
        return self.__add__(other)
        
    def __sub__(self, other):
        if isinstance(other, int):
            other = Frac(other, 1)
        if isinstance(other, Frac):
            new_nom = self.nominator * other.denominator - other.nominator * self.denominator
            new_den = self.denominator * other.denominator
            result = Frac(new_nom, new_den)
            return result.mixed()
        return NotImplemented

    def __rsub__(self, other):
        if isinstance(other, int):
            other = Frac(other, 1)
        if isinstance(other, Frac):
            new_nom = other.nominator * self.denominator - self.nominator * other.denominator
            new_den = self.denominator * other.denominator
            result = Frac(new_nom, new_den)
            return result.mixed()
        return NotImplemented

    def __mul__(self, other):
        if isinstance(other, int):
            result = Frac(self.nominator * other, self.denominator)
            return result.mixed()
        if isinstance(other, Frac):
            return NotImplemented
        
    def __rmul__(self, other):
        return self.__mul__(other)

    def __truediv__(self, other):
        if isinstance(other, int):
            other = Frac(other, 1)
        if isinstance(other, Frac):
            new_nom = self.nominator * other.denominator
            new_den = self.denominator * other.nominator
            result = Frac(new_nom, new_den)
            return result.mixed()
        return NotImplemented

    def __rtruediv__(self, other):
        if isinstance(other, int):
            other = Frac(other, 1)
        return other.__truediv__(self)


    def simplify(self, value=None):
        # simplifies to most simple form unless value is given
        if value is None:
            g = gcd(self.nominator, self.denominator) # gcd = Greatest common divider -> int
            self.nominator //= g
            self.denominator //= g
        
            return f"{self.nominator}/{self.denominator}"
        else:
            g = (value - self.denominator) / 2
            self.nominator += g

            return f"{int(self.nominator)}/{value}"

    def __str__(self):
        return f"{self.nominator}/{self.denominator}"
    
    def mixed(self):
        # represent the fraction in mixed terms
        whole = self.nominator // self.denominator
        rest = self.nominator % self.denominator

        if rest == 0:
            return str(whole)
        elif whole == 0:
            return self.simplify()
        else:
            g = gcd(rest, self.denominator)
            rest //= g
            den = self.denominator // g
            return f"{whole} {rest}/{den}"

    def __eq__(self, other):
        # checks equality by overloading ==
        self.other = other
      
        if (self.nominator / self.denominator) == other:
            return True
        else:
            return False

# __eq__ & simplify
a = Frac(2, 4)
b = Frac(1, 2)
if a == b:
    print("It's a match!")
print(a.simplify(8))

# __mul__ & __rmul__
c = Frac(1, 2)
print(c * 3)
print(3 * c)

# mixed & __str__
d = Frac(30, 12)
print(d)
print(d.mixed())
print(d)

# __add__ & __radd__
e = Frac(1, 4)
print(e+2)
print(2+e)

# __sub__ & __rsub__
f = Frac(30,10)
print(f-2)
print(2-f)

# __truediv__ & __rtruediv__
g = Frac(100,10)
print(g / 2)
print(2 / g)
````
