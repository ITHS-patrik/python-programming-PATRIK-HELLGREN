## 1. Check sign (*)

Ask the user to input a number and check if this number is **positive**, **negative** or **zero** and print it.

```python
# Uppgift 1.1
value = int(input("Enter a value: "))

if value == 0:
    print("Value is zero.")
elif value > 0:
    print("Value is positive.")
else:
    print("Value is negative.")
```

## 2. Smallest (*)
Ask the user to input two numbers and check which one is the smallest and print it out.   

```python
# Uppgift 1.2
value1 = int(input("Enter first value: "))
value2 = int(input("Enter second value: "))

if value1 > value2:
    print(f"{value2} is the smallest number.")
elif value1 < value2:
    print(f"{value1} is the smallest number.")
else:
    print("Both numbers are equal.")
```

## 3. Right angle (*)

Ask the user to input three angles and check if the triangle has a right angle. Your code should make sure that all three angles are valid and make up a triangle.

<img align="left" src="../assets/triangle_angles.png" width="200"/>

```python
# Uppgift 1.3
tryAgain = False

while tryAgain == False:
    angle1 = int(input("Enter first angle: "))
    angle2 = int(input("Enter second angle: "))
    angle3 = int(input("Enter third angle: "))

    sum_angles = angle1 + angle2 + angle3
   
    if sum_angles == 180 and angle1 > 0 and angle2 > 0 and angle3 > 0:
        print("The angles form a valid triangle")
        if angle1 == 90 or angle2 == 90 or angle3 == 90:
            print("There is also a right angle in the triangle. Exiting the program.")
            tryAgain = True
        else:
            print("Although, there is no right angle in the triangle. Exiting the program.")
            tryAgain = True
    else:
        print("The angles do not form a valid triangle, try again.")
```

## 4. Medicine (*)

The information in the following table is stated in a medicine package. Also it is stated that for children weight is more important than age.

        <th>Age</th>

  <tr>

    <td>7-12 years</td>
    <td>1/2-1</td>

  <tr>

    <td>3-7 years</td>

  </tr>

Let the user input an **age** and a **weight**, the program should recommend the number of pills for the user.

```python
# Uppgift 1.4

age = int(input("Enter your age: "))
weight = int(input("Enter your weight in kg: "))

if age > 12 or (age <= 12 and weight > 40):
    print("1-2 pills")
elif age <= 12:
    if weight <= 25:
        print("1/2 pill")
    elif weight > 25 and weight <= 40:
        print("1/2-1 pill")
else:
    print("You have entered an invalid age.")
```

## 5. Divisible (*)
Let the user input a number. Check if the number is
<ol type="a">
  <li>even or odd</li>
  <li>is divisible by 5</li>
  <li>is divisble by 5 and odd</li>
</ol> 

```python
# Uppgift 1.5
value = int(input("Enter a value: "))
```

```python
# a
if value % 2 == 0:
    print("Value is even.")
else:
    print("Value is odd.")
```

```python
# b
divisor = 5
if value % divisor == 0:
    print(f"Value is divisible by {divisor}.")
else:
    print(f"Value is not divisible by {divisor}.")
```

```python
# c
if value % 2 == 0 and value % divisor == 0:
    print(f"Value is even and divisible by {divisor}.")
else:
    print(f"Value is not both even and divisible by {divisor}.")
```

## 6. Luggage size (*)

The maximum allowed luggage size for boarding an airplane is: 
- weight: 8kg
- dimensions: 55x40x23cm (length x width x height) 

Let the user input weight, length, width and height of the luggage. The program should check if the luggage is allowed or not.

```python
# Uppgift 1.6
maxWeight = 8
maxLength = 55
maxWidth = 40
maxHeight = 23

weight = int(input("Enter the weight of the luggage in kg: "))
length = int(input("Enter the length of the luggage in cm: "))
width = int(input("Enter the width of the luggage in cm: "))
height = int(input("Enter the height of the luggage in cm: "))

if weight > maxWeight:
    print("Luggage is too heavy")
elif length > maxLength or width > maxWidth or height > maxHeight:
    print("Luggage is too large")
else:
    print("Luggage is acceptable")
```
